Loading each dataset into a pandas DataFrame and checking its shape.

```python
import pandas as pd
```

## Primary 1: Fake news on Twitter, 2016 election (Grinberg et al. 2019)

```python
twitter_daily = pd.read_csv("https://raw.githubusercontent.com/LazerLab/twitter-fake-news-replication/master/figure_1/data/daily_counts_exposures.csv")

print(twitter_daily.shape)
twitter_daily.head()
```

```python
twitter_daily.dtypes
```

```python
print(twitter_daily["date"].min(), "to", twitter_daily["date"].max())
```

```python
import matplotlib.pyplot as plt

twitter_daily['date'] = pd.to_datetime(twitter_daily['date'])
twitter_daily = twitter_daily.sort_values('date')

plt.figure(figsize=(10, 5))
plt.plot(twitter_daily['date'], twitter_daily['pct_fake'])
plt.axvline(pd.to_datetime('2016-11-08'), color='red', linestyle='--', label='Election Day')
plt.title('Share of Twitter News Views from Fake News Sites, Aug-Dec 2016')
plt.xlabel('Date')
plt.ylabel('Fraction of news views from fake news')
plt.xticks(rotation=45)
plt.legend()
plt.show()
```

```python
twitter_sites = pd.read_csv("https://raw.githubusercontent.com/LazerLab/twitter-fake-news-replication/master/coexposure_network/data/total_exposure_counts.csv")

print(twitter_sites.shape)
twitter_sites.sort_values("N", ascending=False).head(10)
```

## Primary 2: FakeNewsNet (PolitiFact articles)

```python
fake = pd.read_csv("https://raw.githubusercontent.com/KaiDMML/FakeNewsNet/master/dataset/politifact_fake.csv")
real = pd.read_csv("https://raw.githubusercontent.com/KaiDMML/FakeNewsNet/master/dataset/politifact_real.csv")

fake["label"] = "fake"
real["label"] = "real"
news = pd.concat([fake, real])

print(news.shape)
news.head()
```

```python
news.dtypes
```

```python
# tweet_ids is a list of tweet IDs separated by tabs, so counting them = number of tweets
news["num_tweets"] = news["tweet_ids"].str.split().str.len()

news.groupby("label")["num_tweets"].median()
```

## Secondary 2: News website quality ratings (Lin et al. 2023)

```python
ratings = pd.read_csv("https://raw.githubusercontent.com/hauselin/domain-quality-ratings/main/data/domain_pc1.csv")

print(ratings.shape)
ratings.head()
```

```python
# how many Twitter websites also have a quality rating? (checks that the join will work)
matches = twitter_sites["website"].isin(ratings["domain"]).sum()
print(matches, "out of", len(twitter_sites), "Twitter websites have a quality rating")
```
